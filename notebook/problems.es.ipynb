## Ejercicio 1 

Dos dados se lanzan una vez y se observa el total obtenido. Usa una simulación para encontrar la probabilidad estimada de que la puntuación sea mayor a 7 o un número par. Una simulación es una repetición del mismo experimento multiples veces para observar su comportamiento:

- Ejecuta el experimento 1000 veces (lanza 2 dados 1000 veces, y suma el número de ambos dados).
- Lleva ua cuenta de los números y las veces que la suma fue mayor a 7 o un número par.
- Divide el número del paso 2 entre el número de iteraciones(1000).

```python
# TODO

import random

# Simular "n" lanzamientos de dos dados y calcular la probabilidad de que el sum sea mayor que 7 
# o sea un número par

def roll_the_dice(n):
    
    count = 0   # Inicializamos el contador

    for _ in range (n): # Bucle para simular el lanzamiento de los dados "n" veces

        dice_1 = random.randint(1,6)    # Genera un número aleatorio entre el 1 y el 6 (dice 1)

        dice_2 = random.randint(1,6)    # Genera un número aleatorio entre el 1 y el 6 (dice 2)

        event = dice_1 + dice_2     # Calcular la suma de los dos dados

        if (event > 7) or (event % 2 == 0):
            ''' Condición 1: Si la suma es mayor que 7
                    Condición 2: Si es número par'''

            count += 1     # Si se cumple la condición, incrementamos el contador

    # Obtenemos el número de veces que se cumple la condición (count) entre el número de ejecuciones (lanzamientos)

    return count / n

# Ejecutamos e imprimimos el resultado de la probabilidad redondeando el porcentaje a 2 decimales
print(f"The probability that the score is greater than 7 or an even number is: {round(roll_the_dice(1000)*100, 2)}%")
```

## Ejercicio 2

Una caja contiene 10 bolas blancas, 20 bolas rojas y 30 bolas verdes. Si tomamos 5 bolas de la caja reemplazándolas (tomando una bola, anotando el color y luego regresandola a la caja). Queremos saber la probabilidad de:

1. Tomar 3 blancas y 2 rojas.
2. Toas todas del mismo color.

Ejecuta el experimento 1000 veces y calcula las probabilidades que mencionadas.

```python
ball_box = {}

# Crea la cajas con las bolas
for i in range(60):
    if i < 10:
        ball_box[i] = "White"
    elif (i > 9) and (i < 30):
        ball_box[i] = "Red"
    else:
        ball_box[i] = "Green"

print(ball_box)
            
# TODO

import numpy as np

# Función para sacar bolas de la caja

def experimento (n):

    blancas_y_rojas = 0     # Contador tomar 3 bolas blancas y 2 rojas

    mismo_color = 0     # Contador todas las bolas del mismo color

    # Repetición del experimento 'n' veces

    for _ in range (n):    
        lista = []  # Lista vacía donde se guardarán los colores de cada repetición

        # Seleccionamos 5 bolas de forma aleatoria

        for _ in range(5): 

            # Seleccionamos una bola aleatoriamente entre los índices 0 y 59. Después añadimos su color a la lista
            lista.append(ball_box[np.random.randint(0,59)])

            '''Para optimizar el código también podemos usar
                 np.random.choice(ball_box, 5, replace = False) 
                    que seleciona 5 bolas sin reemplazo
                        y nos ahorrariamos el segundo bucle'''

        blancas = lista.count('White')    # Número de bolas blancas
        rojas = lista.count('Red')        # Número de bolas rojas
        verdes = lista.count('Green')     # Número de bolas verdes

        if (blancas == 3) and (rojas == 2): 
            '''Condición 1: 3 bolas blancas
                            and
                 Condición 2: 2 bolas rojas'''
            blancas_y_rojas += 1  # Incrementamos el contador si se cumplen las condiciones

        if (blancas == 5) or (rojas == 5) or (verdes == 5):
            '''Condición 1: 5 bolas blancas
                                or
                    Condición 2: 5 bolas rojas
                                or
                        Condicón 3: 5 bolas verdes'''
            mismo_color += 1    # Incrementamos el contador si se cumplen las condiciones

   
    return   blancas_y_rojas/n, mismo_color/n  # Retornamos los casos favorables entre los casos posibles

# Ejecutamos el experimento 1000 veces

probabilidades = experimento(1000)

# Imprimimos el valor multiplicado por 100 para obtener el porcentaje de cada probabilidad y lo redondeamos a dos decimales
print(f"The probabilities of getting three white balls and two red balls are: {round(probabilidades[0]*100,2)}%")
print(f"The probabilities of all balls coming out the same color are:{round(probabilidades[1]*100,2)}%")
```
